```python
# scripts/run_workspace.py
import time
from pathlib import Path
from workspace import Workspace
from workspace.recipes.change_tool import ChangeTool
from workspace.recipes.handle_microtube import HandleMicrotube


ws = Workspace(config_path="config/config.yaml")
core = ws.components["core"]
core.simulation(True)

tool_rack_1 = ws.components["tool_rack_1"]
tool_rack_2 = ws.components["tool_rack_2"]

microplate_1 = ws.components["microplate_1"]
microplate_2 = ws.components["microplate_2"]
change_tool_1 = ChangeTool(workspace=ws, core=core, tool_rack=tool_rack_1, speed_factor=.1,left_approach=False)
change_tool_2 = ChangeTool(workspace=ws, core=core, tool_rack=tool_rack_2, speed_factor=.1,left_approach=True,base_distance=150)
handle_microtube_1 = HandleMicrotube(workspace=ws, core=core, microplate=microplate_1, speed_factor=4,left_approach=False)
handle_microtube_2 = HandleMicrotube(workspace=ws, core=core, microplate=microplate_2, speed_factor=4,left_approach=False)
```
