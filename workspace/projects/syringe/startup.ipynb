# init

```python
# scripts/run_workspace.py
from workspace import Workspace
from util import create_recipes

# workspace
workspace = Workspace(config_path=["config/base.j2", "config/layout.j2"])
# core
core = workspace.components["core"]

# simulation off
core.simulation(False)
```

# rail

```python
import time

# rail encoder
core.robot_api.set_axis(index=core.rail_cfg["axis"], 
                        usem=core.rail_cfg["usem"], 
                        usee=core.rail_cfg["usee"], 
                        pprm=core.rail_cfg["pprm"],
                        tprm=core.rail_cfg["tprm"], 
                        ppre=core.rail_cfg["ppre"], 
                        tpre=core.rail_cfg["tpre"])
# sleep
time.sleep(1)

# pid
core.robot_api.set_pid(index=core.rail_cfg["axis"], 
                        p=core.rail_cfg["p"], 
                        i=core.rail_cfg["i"], 
                        d=core.rail_cfg["d"], 
                        duration=core.rail_cfg["duration"], 
                        threshold=core.rail_cfg["threshold"])
# sleep
time.sleep(1)

# rail home
core.robot_api.home_with_stop(index=core.rail_cfg["axis"], val=core.rail_cfg["offset"], dir=-1)
```

# feeder

```python
import time
# feeder
feeder = workspace.components["feeder"]
# encoder
core.robot_api.set_axis(index=feeder.axis_cfg["axis"], 
                        usem=feeder.axis_cfg["usem"], 
                        usee=feeder.axis_cfg["usee"], 
                        pprm=feeder.axis_cfg["pprm"],
                        tprm=feeder.axis_cfg["tprm"], 
                        ppre=feeder.axis_cfg["ppre"], 
                        tpre=feeder.axis_cfg["tpre"])
# sleep
time.sleep(1)
# pid
core.robot_api.set_pid(index=feeder.axis_cfg["axis"], 
                        p=feeder.axis_cfg["p"], 
                        i=feeder.axis_cfg["i"], 
                        d=feeder.axis_cfg["d"], 
                        duration=feeder.axis_cfg["duration"], 
                        threshold=feeder.axis_cfg["threshold"])

# sleep
time.sleep(1)

# home feeder
core.robot_api.home_with_encoder_index(index=feeder.axis_cfg["axis"], val=feeder.axis_cfg["offset"])
```
