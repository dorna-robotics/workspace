```python
positions = [f"{row}{col}" for row in "ABCDEF" for col in range(1, 9)]

for i in range(len(positions)):
    pick = positions[i]
    place = positions[(i+1) % len(positions)]
    rcp["cap_holder"].pick_from(pick)
    rcp["cap_holder"].place_in(place)
```

# decapper

```python
indices = ["A2", "A3", "A4", "A5", "A6", "A7", "A8"]
for index in indices:
    # pick tube
    rcp["sbs_plate"].pick_from(index)
    # place in decapper
    rcp["decapper"].place_tube()
    # pick cap
    rcp["cap_holder"].pick_from(index)
    # capping
    rcp["decapper"].cap(exit=False)
    # pick from decapper
    rcp["decapper"].pick_tube(approach=False)
    # back to sbs plate
    rcp["sbs_plate"].place_in(index)  
```
