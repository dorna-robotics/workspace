# prm

```python
source = [
    ["rack_40ml_1", "A1"],
    ["rack_40ml_1", "A2"],
    ["rack_40ml_1", "A3"],
    ["rack_40ml_1", "A4"],
]

working = [
    ["rack_40ml_2", "B1"],
    ["rack_40ml_2", "B2"],
    ["rack_40ml_2", "B3"],
    ["rack_40ml_2", "B4"],
]

cap_holder = [
    "decapper_1",
    "decapper_2",
    "decapper_3",
    "decapper_4",
]

dosing_40ml = [
    ["doser_40ml_1", "B1"],
    ["doser_40ml_1", "B2"],
    ["doser_40ml_1", "B3"],
    ["doser_40ml_1", "B4"],
]

dosing_clean = [
    ["doser_40ml_1", "A1"],
    ["doser_40ml_1", "A2"],
    ["doser_40ml_1", "A3"],
]

dosing_waste = [
    ["doser_40ml_1", "A4"],
]

shaker = [
    ["shaker_1", "A1"],
    ["shaker_1", "A2"],
    ["shaker_2", "A1"],
    ["shaker_2", "A2"],
]

cap_feeder = [
    ["cap_holder_1", "A1"],
    ["cap_holder_1", "A2"],
    ["cap_holder_1", "A3"],
    ["cap_holder_1", "A4"],
]

dosing_2ml_end = [
    ["doser_2ml_1", "A1"],
    ["doser_2ml_1", "A2"],
    ["doser_2ml_1", "A3"],
    ["doser_2ml_1", "A4"],
]

dosing_2ml_middle = [
    ["doser_2ml_2", "A1"],
    ["doser_2ml_2", "A2"],
    ["doser_2ml_2", "A3"],
    ["doser_2ml_2", "A4"],
]

rack_2ml_end = [
    ["rack_2ml_1", "A1"],
    ["rack_2ml_1", "A2"],
    ["rack_2ml_1", "A3"],
    ["rack_2ml_1", "A4"],
]

rack_2ml_middle = [
    ["rack_2ml_2", "A1"],
    ["rack_2ml_2", "A2"],
    ["rack_2ml_2", "A3"],
    ["rack_2ml_2", "A4"],
]
# dosing prm
immerse_40ml_dist = 90
retract_40ml_dist = 10
immerse_2ml_dist = 25
retract_2ml_dist = 10

# shake
shake_duration = 0

# inspection
inspection_frq = 4
inspection_rot = 90

```

# scale , inspect, and decap

```python
# pick gripper
rcp["tool_rack_1"].pick()

for i in range(len(source)):
    # pick tube
    rcp[source[i][0]].pick_from(source[i][1])

    # place scale
    rcp["scale_1"].place_in("place")
    # weight
    rcp["scale_1"].weight()
    # pick from scale
    rcp["scale_1"].pick_from("place")
    # inspect
    rcp["inspector_1"].present(approach=False)
    for _ in range(inspection_frq):
        rcp["inspector_1"].rotate(rotation=inspection_rot)
    # place in decapper and decap
    rcp["decapper_5"].place(exit=False)
    rcp["decapper_5"].decap(approach=False)
    # place cap
    rcp[cap_holder[i]].place()
    # grab tube and place 
    rcp["decapper_5"].pick()
    rcp[working[i][0]].place_in(working[i][1])
    

# place tool
rcp["tool_rack_1"].place()
```

# dosing 40ml

```python
# pick gripper
rcp["tool_rack_4"].pick()

for i in range(len(source)):
    # dispense
    rcp[dosing_40ml[i][0]].immerse(dist=immerse_40ml_dist, anchor=dosing_40ml[i][1])
    rcp[dosing_40ml[i][0]].dispense(vol=10)
    rcp[dosing_40ml[i][0]].retract(dist=retract_40ml_dist, anchor=dosing_40ml[i][1])

    # clean
    for j in range(len(dosing_clean)):
        # clean tube
        rcp[dosing_clean[j][0]].immerse(dist=immerse_40ml_dist, anchor=dosing_clean[j][1])
        rcp[dosing_clean[j][0]].dispense(vol=10)
        rcp[dosing_clean[j][0]].retract(dist=retract_40ml_dist, anchor=dosing_clean[j][1])
        # leftover tube
        rcp[dosing_waste[0][0]].immerse(dist=immerse_40ml_dist, anchor=dosing_waste[0][1], padding=10)
        rcp[dosing_waste[0][0]].dispense(vol=10)
        rcp[dosing_waste[0][0]].retract(dist=retract_40ml_dist, anchor=dosing_waste[0][1])

# place tool
rcp["tool_rack_4"].place()
```

# capping and shake

```python
# pick gripper
rcp["tool_rack_1"].pick()

for i in range(len(source)):
    # pick tube
    rcp[working[i][0]].pick_from(working[i][1])

    # place tube
    rcp["decapper_5"].place()
    # pick cap
    rcp[cap_holder[i]].pick()
    # place cap
    rcp["decapper_5"].cap(exit=False)
    rcp["decapper_5"].pick(approach=False)

    # shake
    rcp[shaker[i][0]].place_in(shaker[i][1])

# shake
rcp[shaker[i][0]].shake(duration=shake_duration)
    

# place tool
rcp["tool_rack_1"].place()
```

# cap feeder

```python
# pick suction
rcp["tool_rack_3"].pick()

for i in range(len(source)):
    # above the feeder
    rcp["feeder_1"].above(anchor="plate_center")
    # detect and present
    rcp["feeder_1"].present_cap(rcp["inspector_1"])  
    # pick from feeder
    rcp["feeder_1"].pick(approach=False)
    # place in cap holder
    rcp[cap_feeder[i][0]].place_in(cap_feeder[i][1])

# place tool
rcp["tool_rack_3"].place()
```

# shaker to decap

```python
# pick gripper
rcp["tool_rack_1"].pick()

for i in range(len(source)):
    # pick tube
    rcp[shaker[i][0]].pick_from(shaker[i][1])

    # place in decapper and decap
    rcp["decapper_5"].place(exit=False)
    rcp["decapper_5"].decap(approach=False)
    # place cap
    rcp[cap_holder[i]].place()
    # grab tube and place 
    rcp["decapper_5"].pick()
    rcp[working[i][0]].place_in(working[i][1])

# place tool
rcp["tool_rack_1"].place()
```

# dosing 2ml

```python
# pick gripper
rcp["tool_rack_4"].pick()

for i in range(len(source)):
    # dose 40ml
    rcp[dosing_40ml[i][0]].immerse(dist=immerse_40ml_dist, anchor=dosing_40ml[i][1], padding=150)
    rcp[dosing_40ml[i][0]].dispense(vol=10)
    rcp[dosing_40ml[i][0]].retract(dist=retract_40ml_dist, anchor=dosing_40ml[i][1])
    
    # dose 2ml_middle
    rcp[dosing_2ml_middle[i][0]].immerse(dist=immerse_2ml_dist, anchor=dosing_2ml_middle[i][1])
    rcp[dosing_2ml_middle[i][0]].dispense(vol=10)
    rcp[dosing_2ml_middle[i][0]].retract(dist=retract_2ml_dist, anchor=dosing_2ml_middle[i][1])

    # dose 2ml_end
    rcp[dosing_2ml_end[i][0]].immerse(dist=immerse_2ml_dist, anchor=dosing_2ml_end[i][1])
    rcp[dosing_2ml_end[i][0]].dispense(vol=10)
    rcp[dosing_2ml_end[i][0]].retract(dist=retract_2ml_dist, anchor=dosing_2ml_end[i][1])

    # clean
    for j in range(len(dosing_clean)):
        # clean tube
        rcp[dosing_clean[j][0]].immerse(dist=immerse_40ml_dist, anchor=dosing_clean[j][1])
        rcp[dosing_clean[j][0]].dispense(vol=10)
        rcp[dosing_clean[j][0]].retract(dist=retract_40ml_dist, anchor=dosing_clean[j][1])
        # leftover tube
        rcp[dosing_waste[0][0]].immerse(dist=immerse_40ml_dist, anchor=dosing_waste[0][1])
        rcp[dosing_waste[0][0]].dispense(vol=10)
        rcp[dosing_waste[0][0]].retract(dist=retract_40ml_dist, anchor=dosing_waste[0][1])

# place tool
rcp["tool_rack_4"].place()
```

# cap 40ml and return to source

```python
# pick gripper
rcp["tool_rack_1"].pick()

for i in range(len(source)):
    # pick tube
    rcp[working[i][0]].pick_from(working[i][1])

    # place tube
    rcp["decapper_5"].place()
    # pick cap
    rcp[cap_holder[i]].pick()
    # place cap
    rcp["decapper_5"].cap(exit=False)
    rcp["decapper_5"].pick(approach=False)

    # return to source
    rcp[source[i][0]].place_in(source[i][1])
    

# place tool
rcp["tool_rack_1"].place()
```

# cap 2ml and scan and return

```python
# pick gripper
rcp["tool_rack_2"].pick()

for i in range(len(source)):
    # pick tube
    rcp[rack_2ml_end[i][0]].pick_from(rack_2ml_end[i][1])

    # place tube
    rcp["decapper_5"].place()
    # pick cap
    rcp[cap_feeder[i][0]].pick_from(cap_feeder[i][1])
    # place cap
    rcp["decapper_5"].cap(exit=False)
    rcp["decapper_5"].pick(approach=False)

    # inspection
    rcp["inspector_1"].present(approach=False)
    for _ in range(inspection_frq):
        rcp["inspector_1"].rotate(rotation=inspection_rot)

    # return to 2ml
    rcp[rack_2ml_end[i][0]].place_in(rack_2ml_end[i][1])

# place tool
rcp["tool_rack_2"].place()
```
